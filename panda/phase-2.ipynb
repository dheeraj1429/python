```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import matplotlib.image as mpimg
```

```python
x = np.linspace(0, 5, 11)
y = x ** 2
```

```python
x, y
```

```python
plt.plot(x, y)
```

```python
plt.subplot(2,2,1)
plt.plot(x**2, x)
plt.subplot(2,2,4)
plt.plot(x,y)
plt.subplot(2,2,2)
plt.plot(x,y**2)
plt.subplot(2,2,3)
plt.plot(y,x)
```

```python
fig = plt.figure(figsize=(5,5),dpi=100)
axis1 = fig.add_axes([0.1,0.1,0.8,0.8])
axis1.plot(x, y)

axis2 = fig.add_axes([0.2,0.2,0.4,0.4])
axis2.plot(x, y)
```

```python
plt.scatter(x,y)
```

```python
plt.hist(x,y)
```

```python
fig = plt.figure()
ax = fig.add_axes([0.1,0.1,0.8,0.8])
ax.plot(x,y)
plt.savefig("basic")
```
