```python
import re
```

```python
chat = """
you are asking to many questions 👎🏿 1234567891 acb@gmail.com here it is: (1233)-567-879, abc@gmail.uk.com yes, phone: 1234567891 email: abc@gmail.com
"""
```

```python
pattern_for_numbers = '\d{10}|\(\d{3,}\)-\d{3,}-\d{3,}'
user_numbers = re.findall(pattern_for_numbers, chat)
```

```python
pattern_for_email = '[a-zA-Z0-9_/+-.]+@[a-zA-Z0-9_/+-.]{2,}'
user_emails = re.findall(pattern_for_email, chat)
```

```python
combine_result = zip(user_numbers, user_emails)
print(list(combine_result))
```

```python
text = """
The cat sat next to the dog. A catalog was on the desk while dogma was discussed Dog.
"""
print(re.findall(r"\b(cat|dog)\b", text, re.IGNORECASE))
```

```python
text = """
Note 1 - Overview
This is the demo note
Note 2 - This is the second title
This is the demo note.
Note 1232 - This is the second title
This is the demo note.
"""
pattern = 'Note \d+ - ([^\n]+)'
print(re.findall(pattern, text))
```

```python
text = """
The gross cost of  operating lease vechiles in FY2021 Q1 was $4.85 billion.
In previous quarter i.e. FY2020 Q4 it was $3 billion.
FY2020 Q5
fy2021 q4
"""
pattern = 'FY(\d{4} Q[1-4])'
print(re.findall(pattern, text, re.IGNORECASE))
```

```python
text = """
The gross cost of  operating lease vechiles in FY2021 Q1 was $4.85 billion.
In previous quarter i.e. 2450 FY2020 Q4 it was $3 billion.
$123.345 123.345$
123$.1235
FY2020 Q5
"""
pattern = '[(\s|\n|\t)](\$[^\s]+)'
print(re.findall(pattern, text))
```
