```python
price = {
    "a": 10,
    "b": 20,
    "c": 30
}
quntity = {
    "a": 2,
    "c": 1
}
```

```python
results = []
for (key, val) in price.items():
    item = quntity.get(key)
    if(item):
       results.append(item * val) 
print(results)
```
