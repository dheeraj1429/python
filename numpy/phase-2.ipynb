```python
cumsum = np.cumsum(sales_data[:, 1:], axis=1)
plt.figure(figsize=(10, 6))
plt.plot(np.mean(cumsum, axis=0))
plt.title("Average cumulative sales per years")
plt.xlabel("Years")
plt.ylabel("Sales")
plt.grid(True)
plt.show()
```
